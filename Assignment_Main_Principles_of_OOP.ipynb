# Hands On

dari program diatas kita bisa mengubah username menggunakan method get dan set, lalu bagaimana jika saya ingin mengubah gender "Male" dari user Jonathan menjadi "unknown"???

```python
class User:
    def __init__(self, username, gmail, gender, status):
        self._username = username  # Enkapsulasi dengan menggunakan underscore (_)
        self._gmail = gmail
        self._status = status

    def get_username(self):
        return self._username

    def get_gmail(self):
        return self._gmail

    def get_gender(self):
        return self._gender

    def get_status(self):
        return self._status

    def set_username(self, username):
        # Validasi atau logika tambahan dapat ditambahkan di sini
        self._username = username

    def set_gender(self, gender):
        # Validasi atau logika tambahan dapat ditambahkan di sini
        self._gender = gender

    def set_status(self, status):
        # Validasi atau logika tambahan dapat ditambahkan di sini
        self._status = status

# Contoh penggunaan
user = User("Wahyudi Putra M", "wahyudiputr@gmail.com", "Male", "Jomblo")
print(user.get_username())  # Mengakses melalui metode getter
user.set_username("Yudi Saja")  # Mengubah melalui metode setter
print(user.get_username())

print(user.get_gmail())

user.set_gender("Unknown") # Mengubah melalui metode setter
print(user.get_gender())

user.set_status("Unknown")
print(user.get_status())
```

# Hands On

Buatlah program yang mengandung 2/3 Main Principles of OOP selain contoh diatas.
Lalu presentasikan kedepan satu-satu

```python
class Pembayaran:
    def __init__(self, jumlah):
        self.__jumlah = jumlah  # ENKAPSULASI: Atribut private

    def get_jumlah(self):
        return self.__jumlah

    def get_format_rupiah(self):
        # Format angka ke ribuan dengan titik
        return f"{int(self.__jumlah):,}".replace(",", ".")

    def proses_bayar(self):
        pass # Ini akan di-override oleh kelas anak

class TransferBank(Pembayaran): # INHERITANCE
    def proses_bayar(self): # POLYMORPHISM (Overriding)
        print(f"Membayar Rp{self.get_jumlah()} melalui Transfer Bank. Silakan cek aplikasi mobile banking Anda untuk melihat pembayaran wdp x3.")
        print()

class EWallet(Pembayaran): # INHERITANCE
    def proses_bayar(self): # POLYMORPHISM (Overriding)
        print(f"Membayar Rp{self.get_jumlah()} melalui E-Wallet. Scan kode QR di layar untuk melunasi uang CGV ghost in the cell.")
        print()

class Cash(Pembayaran): # INHERITANCE
    def proses_bayar(self): # POLYMORPHISM (Overriding)
        print(f"Membayar Rp{self.get_jumlah()} melalui tatap muka. Ini om uangnya, nanti saya pinjem lagi yah...")
        print()

# Fungsi untuk mendemonstrasikan Polymorphism
def eksekusi_pembayaran(metode):
    print("--- Memulai Transaksi With Iqbal ---")
    metode.proses_bayar()

# Contoh Penggunaan
bayar1 = TransferBank(84000)
bayar2 = EWallet(35000)
bayar3 = Cash(100500)

eksekusi_pembayaran(bayar1)
eksekusi_pembayaran(bayar2)
eksekusi_pembayaran(bayar3)
```
